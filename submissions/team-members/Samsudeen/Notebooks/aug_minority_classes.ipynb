```python
import os
from tensorflow.keras.preprocessing.image import ImageDataGenerator, img_to_array, load_img
from tqdm import tqdm
```

# Addressing Class inbalance at data level

```python
# Paths
train_dir = "C:/Users/OWNER/Downloads/SDS-CP028-smart-leaf/submissions/team-members/Samsudeen/TrainValTestDir/train"
# It strikes a balance between fully eliminating class imbalance and avoiding overfitting due to excessive synthetic data.
# Target image counts per class
target_counts = {
    "Corn___Gray_Leaf_Spot": 850,
    "Corn___Northern_Leaf_Blight": 850,
    "Potato___Early_Blight": 850,
    "Potato___Healthy": 850,
    "Potato___Late_Blight": 850,
    "Rice___Brown_Spot": 850,
    "Rice___Leaf_Blast": 850,
    "Rice___Neck_Blast": 850,
    "Wheat___Brown_Rust": 850,
    "Wheat___Yellow_Rust": 850
}

# Augmentation strategy
augmentor = ImageDataGenerator(
    rotation_range=10,
    #zoom_range=0.3,
    #width_shift_range=0.2,
    #height_shift_range=0.2,
    #shear_range=0.2,
    #horizontal_flip=True,
    fill_mode="nearest"
)

# Loop through each class needing augmentation
for class_name, target in target_counts.items():
    class_dir = os.path.join(train_dir, class_name)
    current_images = [f for f in os.listdir(class_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    current_count = len(current_images)
    needed = target - current_count

    print(f"\n📁 Augmenting '{class_name}': {current_count} → {target} (adding {needed})")

    if needed <= 0:
        continue

    i = 0
    pbar = tqdm(total=needed, desc=f"Generating {class_name}")

    while i < needed:
        img_name = current_images[i % len(current_images)]
        img_path = os.path.join(class_dir, img_name)

        try:
            image = load_img(img_path)
            x = img_to_array(image)
            x = x.reshape((1,) + x.shape)

            for batch in augmentor.flow(
                x,
                batch_size=1,
                save_to_dir=class_dir,
                save_prefix="aug",
                save_format="jpg"
            ):
                i += 1
                pbar.update(1)
                if i >= needed:
                    break
        except Exception as e:
            print(f"Error processing {img_path}: {e}")

    pbar.close()

print("\n✅ Augmentation complete!")
```
